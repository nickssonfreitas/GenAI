## 2. Create Prompt

```python
prompt = """
Tarefa: Responda com precisão, clareza e concisão às perguntas abaixo. Para cada resposta, forneça uma explicação sucinta do raciocínio usado.

1. Conhecimento Geral
Quem foi o 16º presidente dos Estados Unidos e qual foi seu maior feito?
Qual elemento químico tem o símbolo "Hg"?

2. Raciocínio Lógico
Se todos os gatos são mamíferos e alguns mamíferos podem nadar, todos os gatos podem nadar? Justifique.
Complete a sequência lógica: 2, 6, 12, 20, 30, __?

3. Interpretação de Texto
Texto:
"O desenvolvimento sustentável é aquele que atende às necessidades do presente sem comprometer a capacidade das futuras gerações de atenderem às suas próprias necessidades."
O que esse texto sugere sobre o equilíbrio entre desenvolvimento e preservação ambiental?

4. Matemática Básica
Se um carro percorre 120 km em 2 horas, qual é sua velocidade média?
Qual é o resultado da seguinte expressão: (8 × 5) ÷ (4 + 6)?

5. Criatividade e Contexto
Invente uma frase usando as palavras: "tecnologia", "eficiência" e "sustentabilidade".
"""
```
